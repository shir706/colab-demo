# Creative Ideas for Everyday Objects

Enter everyday objects to get **three numbered, practical ideas** for each one.

Run the cells from top to bottom. When prompted, enter your OpenAI API key; the input is hidden. The key is used only in this Colab runtime and is not part of this notebook.

You can try more than one object. Type **quit**, **q**, or **exit** when asked for an object to stop.

```python
# Install the OpenAI Python SDK in this Colab runtime.
%pip install -q openai
```

```python
from getpass import getpass
from openai import OpenAI

# getpass hides what you type. Do not print or save this key.
api_key = getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)

# Keep asking for objects until the user chooses to stop.
while True:
    object_name = input(
        "Name an everyday object (or type 'quit' to stop): "
    ).strip()

    # Accept a few simple ways to end the loop.
    if object_name.lower() in ("quit", "q", "exit"):
        print("Goodbye!")
        break

    if not object_name:
        print("Please enter an object, or type 'quit' to stop.")
        continue

    # Ask for exactly three useful, realistic, and safe ideas.
    response = client.responses.create(
        model="gpt-6-luna",
        instructions=(
            "Suggest exactly three distinct, practical uses for the object. "
            "Number them 1, 2, and 3. Keep each idea realistic and safe."
        ),
        input=f"The everyday object is: {object_name}",
    )

    # Display the three numbered suggestions.
    print(response.output_text)
```
