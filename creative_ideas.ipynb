# Creative Ideas for Everyday Objects

Enter an everyday object and let the OpenAI API suggest **one unusual, practical use** for it.

Run the cells from top to bottom. When prompted, enter your OpenAI API key; the input is hidden. The key is used only in this Colab runtime and is not part of this notebook.

```python
# Install the OpenAI Python SDK in this Colab runtime.
%pip install -q openai
```

```python
from getpass import getpass
from openai import OpenAI

# getpass hides what you type. Do not print or save this key.
api_key = getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)

# Ask for an everyday object.
object_name = input("Name an everyday object: ").strip()
if not object_name:
    raise ValueError("Please enter an everyday object.")

# Ask for exactly one unusual, practical, and safe idea.
response = client.responses.create(
    model="gpt-6-luna",
    instructions=(
        "Suggest exactly ONE unusual but practical use for the object. "
        "Keep it realistic and safe, and explain it in 2-3 short sentences."
    ),
    input=f"The everyday object is: {object_name}",
)

# Show the model's suggestion.
print(response.output_text)
```
